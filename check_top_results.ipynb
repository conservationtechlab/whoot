```python
import os
file_path = '/mnt/restorage/Audiomoth/Raw sound files/2024/RGCB/AB10/B032/20240524-0607_032_continuous/20240527_000002.WAV'

for file_path in set(best_df["file_path"]):
    test = best_df[best_df["file_path"] == file_path]
    test["Begin Time (s)"] = test["offset"]
    test["End Time (s)"] = test["offset"] + 3
    test["Selection"] = 1
    test["Low Freq (Hz)"] = 0
    test['High Freq (Hz)'] = 16_000
    os.makedirs(os.path.dirname(f"results/buow_predictions{file_path}"), exist_ok=True)
    test.to_csv(f"results/buow_predictions{file_path}".replace("WAV", "txt"), sep="\t")
```
